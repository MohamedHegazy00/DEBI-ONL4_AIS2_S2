```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, r2_score
```

```python
x= np.array([[1],[2],[3],[4]])
y= np.array([2,4,6,8])
```

```python
lin_reg = LinearRegression()
lin_reg.fit(x, y)
y_pred = lin_reg.predict(x)
print (lin_reg.coef_, lin_reg.intercept_)
```

```python
#ridge

lin_ridge = Ridge(alpha=1.0)
lin_ridge.fit(x, y)
y_pred_ridge = lin_ridge.predict(x)
print (lin_ridge.coef_, lin_ridge.intercept_)
```

```python
#lasso
lin_lasso = Lasso(alpha=1.0)
lin_lasso.fit(x, y)
y_pred_lasso = lin_lasso.predict(x)
print (lin_lasso.coef_, lin_lasso.intercept_)
```

```python
print ("Linear Regression MSE:", mean_squared_error(y, y_pred))
print ("r2_score:", r2_score(y, y_pred))
print ("===================================================")
print ("Ridge Regression MSE:", mean_squared_error(y, y_pred_ridge))
print ("r2_score:", r2_score(y, y_pred_ridge))
print ("===================================================")
print ("Lasso Regression MSE:", mean_squared_error(y, y_pred_lasso))
print ("r2_score:", r2_score(y, y_pred_lasso))
print ("===================================================")
```

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.scatter(x, y, color='blue', label='Actual')
plt.plot(x, y_pred, color='red', label='Linear Fit')
plt.title('Linear Regression')
plt.legend()

plt.subplot(1, 3, 2)
plt.scatter(x, y, color='blue', label='Actual')
plt.plot(x, y_pred_ridge, color='red', label='Ridge Fit')
plt.title('Ridge Regression')
plt.legend()


plt.subplot(1, 3, 3)
plt.scatter(x, y, color='blue', label='Actual')
plt.plot(x, y_pred_lasso, color='red', label='Lasso Fit')
plt.title('Lasso Regression')
plt.legend()
plt.tight_layout()
plt.show()
```
