```python
import pandas as pd

def automated_stat_analyzer(df, column_name):
    """
    Company Task: Provide a summary report of a specific data variable.
    """
    column_data = df[column_name]
    
    # Check if the column is numerical
    if pd.api.types.is_numeric_dtype(column_data):
        report = {
            "Type": "Numerical",
            "Mean": column_data.mean(),
            "Median": column_data.median(),
            "Standard Deviation": column_data.std()
        }
    else:
        # For categorical: Calculate the Mode
        # .mode() returns a Series, so we take the first element [0]
        report = {
            "Type": "Categorical",
            "Mode": column_data.mode()[0]
        }
        
    return report
```

## Assignment 2: 
  ### The null_handling_strategy Function


#### Scenario: Incoming user data often has missing values.Students must implement a flexible strategy to handle these "Null Values" to prepare data for Machine Learning.
### Requirements:

* Check for null values in the DataFrame.

* Apply a strategy based on parameters: "drop_rows", "fill_mean", or "fill_median" .

* Ensure the function only fills numerical columns when using mean or median.

```python
import pandas as pd

def null_handling_strategy(df, strategy="fill_mean", column_name="Customer_Age"):
    """
    Company Task: Clean a dataset by resolving missing (NaN) values.
    """
    # 1. Check if the column exists in the DataFrame
    if column_name not in df.columns:
        return df

    # 2. Implementation based on the chosen strategy
    if strategy == "drop_rows":
        # Drop rows where the specific column has missing values
        df = df.dropna(subset=[column_name])
        
    elif strategy == "fill_mean":
        # Fill missing values with the average
        mean_val = df[column_name].mean()
        df[column_name] = df[column_name].fillna(mean_val)
        
    elif strategy == "fill_median":
        # Fill missing values with the middle value
        median_val = df[column_name].median()
        df[column_name] = df[column_name].fillna(median_val)
        
    return df
```
