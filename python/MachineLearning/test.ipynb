```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([1,2,3,4])
y=np.array([2,2.8,3.6,4.5])
```

```python
w=0
b=0
alpha = 0.01
num_iterations = 20
```

```python
sse_values = []
for i in range(num_iterations):
   y_pred = w * x + b
   d_w=2/4*np.sum((y_pred-y)*x)
   d_b=2/4*np.sum(y_pred-y)
   w=w-alpha*d_w
   b=b-alpha*d_b
   sse = np.sum((y_pred - y) ** 2)
   sse_values.append(sse)   



```

```python
print ("Final weight:", w)
print ("Final bias:", b)
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(range(num_iterations), sse_values, label='SSE')
plt.xlabel('Iteration')
plt.ylabel('SSE')
plt.title('SSE vs Iteration')
plt.legend()

plt.subplot(1, 2, 2)
plt.scatter(x, y, color='blue', label='Data Points')
plt.plot(x, w * x + b, color='red', label='Regression Line')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression Fit')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
class LinearRegression:
    def __init__(self, learning_rate=0.01, num_iterations=1000):
        self.learning_rate = learning_rate
        self.num_iterations = num_iterations
        self.w = 0
        self.b = 0

    def fit(self, x, y):
        n = len(x)
        for i in range(self.num_iterations):
            y_pred = self.w * x + self.b
            d_w = (2/n) * np.sum((y_pred - y) * x)
            d_b = (2/n) * np.sum(y_pred - y)
            self.w -= self.learning_rate * d_w
            self.b -= self.learning_rate * d_b

    def predict(self, x):
        return self.w * x + self.b
    def visualize(self, x, y):
        plt.scatter(x, y, color='blue', label='Data Points')
        plt.plot(x, self.predict(x), color='red', label='Regression Line')
        plt.xlabel('x')
        plt.ylabel('y')
        plt.title('Linear Regression Fit')
        plt.legend()
        plt.show()
```
